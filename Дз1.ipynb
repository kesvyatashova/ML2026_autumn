## 2. Теоретическая справка (кратко)
Любую матрицу $ A $ размером $ m \times n $ можно разложить на три матрицы:
$
A = U \cdot \Sigma \cdot V^T
$
Где:
- $ U $ — ортогональная матрица размера $ m \times m $ (левые сингулярные векторы).
- $ \Sigma $ — диагональная матрица размера $ m \times n $, на диагонали которой стоят сингулярные числа $ \sigma_1 \ge \sigma_2 \ge ... \ge \sigma_k $.
- $ V^T $ — ортогональная матрица размера $ n \times n $ (правые сингулярные векторы).

**Идея сжатия:** Если оставить только первые $ r $ сингулярных чисел (и соответствующие столбцы U и V), мы получим матрицу $ A_r $, которая является наилучшим приближением $ A $ среди всех матриц ранга $ r $ (в смысле нормы Фробениуса).

````python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from sklearn.metrics import mean_squared_error

img=Image.open('/content/Lenna_(test_image).png')

img_gray=img.convert('L')

img_array=np.array(img_gray)

original_shape=img_array.shape

plt.imshow(img_array,cmap='gray')
plt.title('Оригинальное изображение')
plt.axis('off')
plt.show()
````

Загрузка и подготовка данных выполнена. Изображение Lena размером 512×512 загружено, переведено в оттенки серого и преобразовано в массив NumPy для дальнейшей обработки методом SVD.

````python
U, s, Vt = np.linalg.svd(img_array, full_matrices=False)
list_of_ranks = [1,5,10,30,100,min(img_array.shape)]

approximations=[]

for r in list_of_ranks:
    reconstructed=(U[:, :r] * s[:r]) @ Vt[:r, :]

    reconstructed=np.clip(reconstructed,0,255)

    approximations.append(reconstructed)
````

Для изображения выполнено восстановление. Получены приближения различного ранга: 1, 5, 10, 30, 100 и полный ранг. При увеличении ранга качество восстановления должно улучшаться.

````python
height, width=img_array.shape

original_size=height*width

original_flat = img_array.flatten()

compression_ratios=[]

for r in list_of_ranks:

    compressed_size=r*(height+1+width)*4

    compression_ratio=original_size/compressed_size

    compression_ratios.append(compression_ratio)
````

Для каждого ранга вычислен коэффициент сжатия на основе размеров матриц. При малых рангах достигается более сильное сжатие, так как требуется хранить значительно меньше данных по сравнению с исходным изображением.

````python
plt.figure(figsize=(12, 8))

for i, r in enumerate(list_of_ranks):

    plt.subplot(2,3,i+1)

    plt.imshow(approximations[i],cmap='gray')

    plt.title(
        f'Ранг = {r}, Сжатие в {compression_ratios[i]:.2f} раз'
    )

    plt.axis('off')

plt.tight_layout()

plt.show()
````

На графиках представлены восстановленные изображения различных рангов. При ранге 1 изображение содержит только самые общие контуры. С увеличением ранга постепенно восстанавливаются детали и улучшается качество изображения. При больших значениях ранга восстановленное изображение становится визуально близким к исходному.
